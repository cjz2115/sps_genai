 ## Question 1

### a) A ∩ B (intersection): “A AND B” - both events occur simultaneously Set interpretation: elements that belong to both set A and set B

```python
p_a = 0.4
p_b = 0.3
```

```python
p_a_and_b = p_a * p_b
print(p_a_and_b)
```

### b) A ∪ B (union): “A OR B” - at least one of the events occurs Set interpretation: elements that belong to either set A or set B (or both)

```python
p_a_or_b = p_a + p_b - p_a_and_b
```

```python
print(p_a_or_b)
```

# Question 3

```python
p_a = 0.6
p_b_given_a = 0.5
p_b_given_not_a = 0.2
```

```python
p_not_a = 1 - p_a
```

```python
p_b = p_b_given_a * p_a + p_b_given_not_a * p_not_a
```

```python
p_a_given_b = p_b_given_a * p_a / p_b
print(p_a_given_b)
```

# Question 4

```python
p_disease = 0.02
p_positive_disease = 0.95
p_positive_healthy = 0.10
```

```python
probability = (p_positive_disease * p_disease) / (
    p_positive_disease * p_disease +
    p_positive_healthy * (1 - p_disease)
)
```

```python
print(probability)
print(probability * 100)
```
